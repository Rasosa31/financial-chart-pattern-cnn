```python
# Data Generation and Dataset Structure

This notebook describes how the dataset used in this project is organized and loaded.
The data consists of financial chart images labeled according to future price movement.

The goal of this step is to:
- Verify dataset integrity
- Understand class distribution
- Prepare the data pipeline for training
```

```python
import os
import matplotlib.pyplot as plt
from PIL import Image
```

```python
BASE_DIR = "data"

TRAIN_DIR = os.path.join(BASE_DIR, "train", "images")
VAL_DIR   = os.path.join(BASE_DIR, "val", "images")
TEST_DIR  = os.path.join(BASE_DIR, "test", "images")
```

```python
def list_classes(directory):
    return sorted([
        d for d in os.listdir(directory)
        if os.path.isdir(os.path.join(directory, d))
    ])

print("Train classes:", list_classes(TRAIN_DIR))
print("Validation classes:", list_classes(VAL_DIR))
print("Test classes:", list_classes(TEST_DIR))
```

```python
def count_images(directory):
    counts = {}
    for cls in list_classes(directory):
        cls_path = os.path.join(directory, cls)
        counts[cls] = len(os.listdir(cls_path))
    return counts

print("Train distribution:", count_images(TRAIN_DIR))
print("Validation distribution:", count_images(VAL_DIR))
print("Test distribution:", count_images(TEST_DIR))
```

```python
def show_examples(directory, n=3):
    classes = list_classes(directory)
    plt.figure(figsize=(10, 4))

    for i, cls in enumerate(classes):
        cls_path = os.path.join(directory, cls)
        images = os.listdir(cls_path)[:n]

        for j, img_name in enumerate(images):
            img_path = os.path.join(cls_path, img_name)
            img = Image.open(img_path)

            plt.subplot(len(classes), n, i*n + j + 1)
            plt.imshow(img)
            plt.axis("off")
            plt.title(cls)

    plt.tight_layout()
    plt.show()

show_examples(TRAIN_DIR)
```

```python
## Summary

- The dataset consists of labeled financial chart images.
- Classes are inferred from directory names.
- There is a noticeable class imbalance.
- Images are visually similar, making the classification task non-trivial.

These observations will be important when interpreting model performance later.
```
