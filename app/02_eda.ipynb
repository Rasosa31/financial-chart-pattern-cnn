```python
# Exploratory Data Analysis (EDA)

In this notebook, we explore the dataset used to train the CNN model.
The objective is to understand:

- Class balance
- Visual similarity between classes
- Potential challenges for image-based classification

This analysis helps interpret model performance results later in the project.
```

```python
import os
import matplotlib.pyplot as plt
import numpy as np
from PIL import Image
```

```python
BASE_DIR = "data"
TRAIN_DIR = os.path.join(BASE_DIR, "train", "images")
```

```python
classes = sorted(os.listdir(TRAIN_DIR))

class_counts = {
    cls: len(os.listdir(os.path.join(TRAIN_DIR, cls)))
    for cls in classes
}

class_counts
```

```python
plt.figure(figsize=(6, 4))
plt.bar(class_counts.keys(), class_counts.values())
plt.title("Class Distribution (Training Set)")
plt.xlabel("Class")
plt.ylabel("Number of Images")
plt.show()
```

```python
total = sum(class_counts.values())

for cls, count in class_counts.items():
    print(f"{cls}: {count/total:.2%}")
```

```python
def show_side_by_side(class_a, class_b, n=5):
    plt.figure(figsize=(10, 4))

    for i in range(n):
        img_a = Image.open(
            os.path.join(TRAIN_DIR, class_a, os.listdir(os.path.join(TRAIN_DIR, class_a))[i])
        )
        img_b = Image.open(
            os.path.join(TRAIN_DIR, class_b, os.listdir(os.path.join(TRAIN_DIR, class_b))[i])
        )

        plt.subplot(2, n, i + 1)
        plt.imshow(img_a)
        plt.axis("off")
        plt.title(class_a)

        plt.subplot(2, n, n + i + 1)
        plt.imshow(img_b)
        plt.axis("off")
        plt.title(class_b)

    plt.tight_layout()
    plt.show()

show_side_by_side("down", "up")
```

```python
def image_intensity_stats(directory, n=100):
    values = []
    for cls in os.listdir(directory):
        cls_path = os.path.join(directory, cls)
        for img_name in os.listdir(cls_path)[:n]:
            img = np.array(Image.open(os.path.join(cls_path, img_name)).convert("L"))
            values.append(img.mean())
    return np.mean(values), np.std(values)

mean_intensity, std_intensity = image_intensity_stats(TRAIN_DIR)
mean_intensity, std_intensity
```

```python
## EDA Conclusions

- The dataset is imbalanced, with one dominant class.
- Financial chart images from different classes are visually very similar.
- No obvious low-level visual features clearly separate the classes.
- This makes the classification task particularly challenging for a CNN.

These findings help explain the instability and class collapse observed during model evaluation.
```
